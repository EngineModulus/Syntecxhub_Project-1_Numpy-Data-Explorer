```python
# # Project 1 — NumPy Data Explorer
#
# This project covers array creation, indexing, slicing, mathematical and statistical operations, reshaping, broadcasting, saving/loading, and performance comparison with Python lists.

import numpy as np
import time
np.set_printoptions(precision=2, suppress=True)


# ## 1. Array creation, indexing and slicing

# Dataset: rows = observations; columns = Age, Height_cm, Weight_kg, Study_Hours
data = np.array([
    [21, 172, 68, 8],
    [23, 180, 75, 6],
    [20, 165, 60, 10],
    [25, 176, 82, 5],
    [22, 169, 64, 9],
    [24, 183, 79, 7],
    [21, 171, 67, 11],
    [26, 178, 85, 4]
], dtype=float)

print("Shape:", data.shape)
print("First row:", data[0])
print("Weight column:", data[:, 2])
print("First five observations:\n", data[:5])
print("Age and height for first three observations:\n", data[:3, :2])


# ## 2. Mathematical and axis-wise operations

ages = data[:, 0]
heights = data[:, 1]
weights = data[:, 2]
study_hours = data[:, 3]

print("Total study hours:", np.sum(study_hours))
print("Mean age:", np.mean(ages))
print("Mean height:", np.mean(heights))
print("Mean weight:", np.mean(weights))
print("SD of weight:", np.std(weights, ddof=1))
print("Minimum/maximum weight:", np.min(weights), np.max(weights))

print("\nColumn means:", np.mean(data, axis=0))
print("Column SDs:", np.std(data, axis=0, ddof=1))
print("Row totals:", np.sum(data, axis=1))


# ## 3. Reshaping and broadcasting

matrix = np.arange(1, 13).reshape(3, 4)
print("Original matrix:\n", matrix)

adjustment = np.array([1, 2, 3, 4])
broadcast_result = matrix + adjustment
print("\nAfter broadcasting:\n", broadcast_result)

column_mean = np.mean(data, axis=0)
column_sd = np.std(data, axis=0, ddof=1)
standardized = (data - column_mean) / column_sd
print("\nStandardized data:\n", standardized)


# ## 4. Statistical summaries

q1, median, q3 = np.percentile(weights, [25, 50, 75])
iqr = q3 - q1

print("Weight statistics")
print("Mean:", np.mean(weights))
print("Median:", median)
print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)

correlation_matrix = np.corrcoef(data, rowvar=False)
print("\nCorrelation matrix:\n", correlation_matrix)


# ## 5. Save and load NumPy arrays

np.save("numPy_data.npy", data)
loaded_data = np.load("numPy_data.npy")

print("Loaded successfully:", np.array_equal(data, loaded_data))


# ## 6. NumPy versus standard Python lists

n = 1_000_000
a_list = list(range(n))
b_list = list(range(n))
a_array = np.arange(n)
b_array = np.arange(n)

start = time.perf_counter()
list_result = [a + b for a, b in zip(a_list, b_list)]
list_time = time.perf_counter() - start

start = time.perf_counter()
array_result = a_array + b_array
numpy_time = time.perf_counter() - start

print(f"Python list time: {list_time:.6f} seconds")
print(f"NumPy array time: {numpy_time:.6f} seconds")
print(f"NumPy speed-up: {list_time / numpy_time:.2f}x")


# ## Conclusion
#
# NumPy provides efficient multidimensional arrays and vectorized operations for numerical computing. Indexing and slicing allow efficient extraction of observations and variables, axis-wise functions provide concise statistical summaries, and reshaping/broadcasting support efficient computation without explicit loops. NumPy also provides convenient array persistence and generally performs numerical operations faster than standard Python lists.
```
